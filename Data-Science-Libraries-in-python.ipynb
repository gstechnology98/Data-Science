<h3 style = "text-align:center;">Adding Title and Lines in Python</h4>

```python
plt.plot(x,y)
plt.title("X VS Y")
plt.xlabel("X-label")
plt.ylabel("Y-label")
plt.show()
```

<h3 style = "text-align:center;">Changing Line Aesthetics in Python</h3>

```python
plt.plot(x, y, color='g', linestyle=':', linewidth=5)
plt.title("X vs Y")
plt.xlabel("This is x axis")
plt.ylabel("This is y axis")
plt.show()
```

<h3 style="text-align:center">Change two lines in the same Plot of line in Python</h3>

```python
x
```

```python
y1 = 2*x
```

```python
y2 = 3 * x
```

```python
plt.plot(x,y1)
plt.plot(x,y2)
plt.show()
```

```python
plt.plot(x,y1,color="g", linestyle=":")
plt.plot(x,y2,color="r", linestyle=":")
plt.title("Two lines in a plot")
plt.xlabel("x-axis")
plt.ylabel("y-label")
plt.grid(True)
plt.show()
```

<h3 style="text-align:center;">Adding sub-plots of Line Plot in Python</h3>

```python
plt.subplot(2,1,1)
plt.plot(x,y1,color="g", linestyle=":")
plt.grid(True)


plt.subplot(2,1,2)
plt.plot(x,y2,color="r", linestyle=":")


plt.title("Two lines in a plot")
plt.xlabel("x-axis")
plt.ylabel("y-label")
plt.grid(True)
plt.show()
```

<h2 style="text-align:center;">Adding Sub-plots of scatter Plot in Python</h2>

```python
x = [1,2,5,7,4,6,8,9,3,10]
y = [9,8,1,2,4,5,3,6,7,12]
z = [3,5,6,8,9,4,1,0,2,7]

plt.subplot(1,2,1)
plt.scatter(x,y,marker="*", c="g",s=100)

plt.subplot(1,2,2)
plt.scatter(x,z,marker="*", c="r",s=200)
plt.show()
```
