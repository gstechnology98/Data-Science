```python
#Loading the required libraries
import pandas as pd
from matplotlib import pyplot as plt 
import seaborn as sns 
```

```python
#loading the npl matches database
npl = pd.read_csv('matches.csv')
```

```python
#heaving a glance at the first five records of the dataset
npl.head()
```

```python
#looking at the member of rows and columns in the dataset
npl.shape
```

```python
#Getting the frequency of most man of the match awards
npl['Players of the match'].value_counts()
```

```python
#Getting the top 10 players with most man of the match awards
npl['Players of the match'].value_counts()[0:10]
```

```python
#Getting the top 5 players with most man of the match awards
npl['Players of the match'].value_counts()[0:5]
```

```python
#Getting the list of top 10 players with most man of the match awards
list(npl['Players of the match'].value_counts()[0:10].keys())
```

```python
plt.figure(figsize=(8,5))
plt.bar(list(npl['Players of the match'].value_counts()[0:5].keys()),list(npl['Players of the match'].value_counts()[0:5]),color="g")
plt.show()
```

```python
#Making a bar-plot for the top 5 players with most man of the match awards
plt.figure(figsize=(8,5))
plt.bar(list(npl['Players of the match'].value_counts()[0:5].keys()),list(npl['Players of the match'].value_counts()[0:5]),color="g")
plt.show()
```

```python
#Getting the frequency of results column
npl['Result'].value_counts()
```

```python
#Finding out the number of toss wins w. r. t. each team
npl['Toss Winner'].value_counts()
```

```python
#Extracting the records where a team won batting first
batting_first=npl[npl['Win by Runs']!=0]
```

```python
#Looking at the head
batting_first.head()
```

```python
#Making a histogram
plt.figure(figsize=(5,5))
plt.hist(batting_first['Win by Runs'])
plt.xlabel("Runs")
plt.show()
```

```python
#Finding out the number of wins w. r. t each after batting first
batting_first['Winner'].value_counts()
```

```python
#Making a bar plot for top 3 teams with most wins after batting first
plt.figure(figsize=(6,6))
plt.bar(list(batting_first['Winner'].value_counts()[0:3].keys()),list(batting_first['Winner'].value_counts()[0:3]),color=["blue", "red", "yellow", "orange", "pink"])
plt.show()
```

```python
#Making a pie chart
plt.figure(figsize=(6,6))
plt.pie(list(batting_first['Winner'].value_counts()),labels = list(list(batting_first['Winner'].value_counts().keys())),autopct='%0.2f%%')
plt.show()
```

```python
#Extracting those records where a team won after batting second
batting_second=npl[npl['Win by Wickets']!=0]
```

```python
batting_second.head()
```

```python
#Making a bar plot for top 3 teams with most wins after batting first
plt.figure(figsize=(6,6))
plt.hist(batting_second['Win by Wickets'],bins=30)
plt.show()
```

```python
#Finding out the frequency of number of wins w. r. t each time after batting second
batting_second['Winner'].value_counts()
```

```python
#Making a bar plot for top-3 team with most wins after batting second
plt.figure(figsize=(7,7))
#plt.bar(list(batting_first['Winner'].value_counts()[0:3].keys()),list(batting_first['Winner'].value_counts()[0:3]),color=["blue", "red", "yellow"])
plt.bar(list(batting_second['Winner'].value_counts()[0:4].keys()),list(batting_second['Winner'].value_counts()[0:4]), color=["blue", "red", "yellow"])
plt.show()
```

```python
#Making a pie chart for distribution of wins after batting second
plt.figure(figsize=(6,6))
plt.pie(list(batting_second['Winner'].value_counts()),labels = list(list(batting_second['Winner'].value_counts().keys())),autopct='%0.2f%%')
plt.show()
```
