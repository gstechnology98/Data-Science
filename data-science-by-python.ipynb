<h3 style="text-align: center;">Over-riding init Method in Python</h3>

```python
class Car(Vehicle):
    def __init__(self,mileage,cost,tyres,hp):
        super().__init__(mileage,cost)
        self.tyres = tyres
        self.hp = hp
    def show_car_details(self):
        print("Number of tyres in car: ", self.tyres)
        print("Horse power of car is ", self.hp)
        print("I am a Car")
```

```python
c1 = Car(45,450000,8,789000)
```

```python
c1.show_car_details()
```

```python
c1.show_vehicle_details()
```

<h1 style="text-align: center;">Multiple Inheritance in Python</h1>

```python
class Parent1:
    def assign_string_one(self,str1):
        self.str1 = str1
    def show_string_one(self):
        return self.str1
```

```python
class Parent2:
    def assign_string_two(self,str2):
        self.str2 = str2
    def show_string_two(self):
        return self.str2
```

```python
class Child(Parent1, Parent2):
    def assign_string_three(self,str3):
        self.str3 = str3
    def show_string_three(self):
        return self.str3
```

```python
my_child=Child()
```

```python
my_child.assign_string_one("I am string of Parent 1")
```

```python
my_child.assign_string_two("I am string of Parent 2")
```

```python
my_child.assign_string_three("I am string of Child")
```

```python
my_child.show_string_one()
```

```python
my_child.show_string_two()
```

```python
my_child.show_string_three()
```

<h1 style="text-align: center;">Multi-level Inheritance in Python</h1>

```python
class Parent:
    def get_name(self,name):
        self.name = name
    def show_name(self):
        return self.name
```

```python
class Child(Parent):
    def get_age(self, age):
        self.age = age
    def show_age(self):
        return self.age
```

```python
class GrandChild(Child):
    def get_gender(self,gender):
        self.gender = gender
    def show_gender(self):
        return self.gender
```

```python
my_gender = GrandChild()
```

```python
my_gender.get_name("Lokendra")
```

```python
my_gender.get_age(45)
```

```python
my_gender.get_gender("Male")
```

```python
my_gender.show_name()
```

```python
my_gender.show_age()
```

```python
my_gender.show_gender()
```
