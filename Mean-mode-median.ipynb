```python
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns 
import datetime as dt 
import numpy as np 
np.random.seed(1)
#Generate random numbers and store it in a varialbe called data
data = np.random.uniform(1,10,20)
data

```

```python
#Calculate Mean
mean = np.mean(data)
print(mean)
```

```python
#Calculate Median
media = np.median(data)
print(media)
```

```python
# Create a function for calculating Mode
# mode <- function(x){
#   ux <- unique(x)
#   ux[which.max(tabulate(match(x, ux)))]
# }

# result <- mode(data)
# print(data)
from collections import Counter

def mode(data):
    return Counter(data).most_common(1)[0][0]

result = mode(data)
print(data)


```

```python
print(result)
```

```python
#Calculate variance and std Deviation
variance = np.var(data)
standardDeviation = np.sqrt(variance)
print(standardDeviation)
```
